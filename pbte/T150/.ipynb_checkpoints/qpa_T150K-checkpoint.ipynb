```python
from ase.io import read
import numpy as np

# 1. Load the POSCAR
# This gives you an Atoms object which knows positions, masses, and cell
atoms = read('POSCAR')
n_atoms = len(atoms)
masses = atoms.get_masses()  # in amu
cell = atoms.get_cell()      # lattice vectors
positions = atoms.get_positions() # Cartesian coordinates
sc_dims = np.array([4, 4, 4])

print(f"Loaded {n_atoms} atoms.")
print(f"{atoms}")
```

```python
import numpy as np
from scipy.interpolate import interp1d

# --- 1. Load BTE.omega (Assumed Band-major after flattening) ---
# data_omega shape: (Nq_irr, num_bands) -> e.g., (455, 6)
data_omega = np.loadtxt('BTE.omega') 
Nq = data_omega.shape[0]
num_bands = data_omega.shape[1]

# Replicate your current logic: Band 1 (all q), then Band 2 (all q)...
# Dividing by 2*pi to convert from rad/ps to THz
w_bare = (data_omega.T).flatten() / (2 * np.pi)
num_total_modes = len(w_bare)

print(f"Total modes: {num_total_modes} from BTE.omega")
print(f"Detected Nq (q-points): {Nq}, Bands: {num_bands}")

# --- 2. Load and Parse BTE.selfeng3ph_spectrum (q-major order) ---
# Each block in this file follows: q1_b1, q1_b2... q1_b6, q2_b1...
spec_data = np.genfromtxt('BTE.selfeng3ph_spectrum')

# Calculate the number of frequency grid points per mode block
pts_per_mode = len(spec_data) // num_total_modes

# Initialize lists to store interpolation functions in Band-major order
d_interps = [None] * num_total_modes
g_interps = [None] * num_total_modes

print(f"Parsing blocks and re-mapping q-major file to Band-major w_bare...")

for i in range(num_total_modes):
    # 'i' represents the sequence in the FILE (q-major)
    # q_idx changes slowly, b_idx changes fast
    q_idx = i // num_bands
    b_idx = i % num_bands
    
    # Calculate the 'target_mode_idx' to match the Band-major 'w_bare'
    # In Band-major: b_idx changes slowly, q_idx changes fast
    target_mode_idx = b_idx * Nq + q_idx
    
    # Extract the data block for the current mode
    start = i * pts_per_mode
    end = (i + 1) * pts_per_mode
    mode_block = spec_data[start:end]
    
    w_grid = mode_block[:, 0] # Frequency grid (THz)
    # Convert Delta and Gamma from rad/ps to THz
    d_grid = mode_block[:, 1] / (2 * np.pi) 
    g_grid = mode_block[:, 2] / (2 * np.pi)
    
    # Map the interpolation function to the correct index for w_bare[target_mode_idx]
    d_interps[target_mode_idx] = interp1d(w_grid, d_grid, kind='linear', fill_value="extrapolate")
    g_interps[target_mode_idx] = interp1d(w_grid, g_grid, kind='linear', fill_value="extrapolate")

print("Successfully loaded and re-ordered all interpolation blocks!")
```

```python
import numpy as np
from scipy.interpolate import interp1d

# --- 1. Load BTE.qpoints to get degeneracies ---
q_data = np.genfromtxt('BTE.qpoints')
# Column 2 (index 2) is the degeneracy (multiplicity)
degeneracies = q_data[:, 2] 
Nq_irr = len(degeneracies) # the number of qpoints in irreducible wedge
Nq_total = np.sum(degeneracies) # the number of qpoints in whole BZ
q_coords = q_data[:, 3:6]

print(f"Number of irreducible q-points: {Nq_irr}")
print(f"Total q-points in full BZ (Sum of weights): {Nq_total}")
```

```python
import numpy as np

# --- Constants ---
kB = 8.617333262e-5       # Boltzmann constant in eV/K
THz_to_eV = 0.004135667   # Conversion factor (h * 1e12) to eV
T = 150.0                 # Temperature in Kelvin

# Fixed Anharmonic Parameters
delta_fixed = 0.0         # Frequency shift (Delta)
gamma_fixed = 1.0         # Broadening (Gamma) in THz

total_cv_weighted = 0.0
frequencies_qpa = []
cv_contributions_qpa = []
integrated_modes_count = 0

for i in range(num_total_modes):
    # 1. Map global index to irreducible q-point index
    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]
    w_qs = w_bare[i]
    
    # Skip unstable or extremely low frequency modes
    #if w_qs < 1e-10: continue

    # 2. Setup Integration Grid for Spectral Density
    g_val = gamma_fixed
    d_val = delta_fixed
    
    # Use adaptive step size
    dw = g_val / 10.0 
    
    # Integration range: +/- 10*Gamma around the shifted peak
    w_min = max(0.001, w_qs + d_val - 100 * g_val)
    w_max = w_qs + d_val + 100 * g_val
    local_grid = np.arange(w_min, w_max, dw)
    
    mode_integral = 0.0
    norm_sum = 0.0  # Used for Spectral Density Normalization
    
    # 3. Numerical Integration
    for w in local_grid:
        # Lorentzian Spectral Density: b_qs(w)
        # Note: (1/pi) * (gamma / ( (w-w_0)^2 + gamma^2 ))
        denom = (w - w_qs - d_val)**2 + g_val**2
        b_w = (1.0 / np.pi) * (g_val / denom)
        
        # Heat Capacity Kernel
        x = (w * THz_to_eV) / (kB * T)
        
        if x < 50:
            cv_kernel = kB * (x**2) * np.exp(x) / (np.exp(x) - 1.0)**2
            
            # Summing values for the integral
            mode_integral += b_w * cv_kernel * dw
            norm_sum += b_w * dw
    
    # 4. Normalize the Spectral Density Integral
    # Even with a +/- 10*Gamma range, the Lorentzian integral is ~0.937.
    # We divide by norm_sum to ensure the total probability is exactly 1.0.
    #if norm_sum > 0:
    #   mode_integral /= norm_sum
    
    # 5. Accumulate Results
    total_cv_weighted += mode_integral * weight
    integrated_modes_count += 1

    frequencies_qpa.append(w_qs)
    cv_contributions_qpa.append(mode_integral)

# --- 6. Final Normalization and Unit Conversion ---
# Normalize by the total number of q-points in the Brillouin Zone
cv_per_cell = total_cv_weighted / Nq_total

# Convert from eV/K/cell to J/K/mol
# (1.60218e-19 J/eV * 6.02214e23 / mol)
cv_j_mol = cv_per_cell * 1.60218e-19 * 6.02214e23

print(f"Total modes integrated: {integrated_modes_count}")
print(f"BQPA Cv (Gamma=0.1, Normalized): {cv_j_mol:.4f} J/K/mol")
```

```python
import numpy as np

# --- Parameters ---
kB = 8.617333262e-5
THz_to_eV = 0.004135667
T = 150.0

# --- Gamma Proportionality Constant ---
# Gamma = alpha * (w_bare^2)
alpha = 1.0 
delta_fixed = 0.0

total_cv_weighted = 0.0
frequencies_qpa = []
cv_contributions_qpa = []

for i in range(num_total_modes):
    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]
    w_qs = w_bare[i]
    
    # 1. Frequency-dependent Gamma 
    g_val = max(alpha * (w_qs**2), 1e-4) 
    d_val = delta_fixed
    
    # 2. Setup Integration Grid
    dw = g_val / 10.0 
    w_min = max(0.001, w_qs + d_val - 100 * g_val)
    w_max = w_qs + d_val + 100 * g_val
    local_grid = np.arange(w_min, w_max, dw)
    
    mode_integral = 0.0
    norm_sum = 0.0
    
    # 3. Numerical Integration
    for w in local_grid:
        denom = (w - w_qs - d_val)**2 + g_val**2
        b_w = (1.0 / np.pi) * (g_val / denom)
        
        x = (w * THz_to_eV) / (kB * T)
        if x < 50:
            cv_kernel = kB * (x**2) * np.exp(x) / (np.exp(x) - 1.0)**2
            mode_integral += b_w * cv_kernel * dw
            norm_sum += b_w * dw
    
    # 4. Normalization 
    #if norm_sum > 0:
    #    mode_integral /= norm_sum
    
    total_cv_weighted += mode_integral * weight
    frequencies_qpa.append(w_qs)
    cv_contributions_qpa.append(mode_integral)

# --- 5. Final Calculation ---
cv_per_cell = total_cv_weighted / Nq_total
cv_j_mol = cv_per_cell * 1.60218e-19 * 6.02214e23

print(f"Alpha: {alpha}")
print(f"BQPA Cv (Gamma prop. to w^2): {cv_j_mol:.4f} J/K/mol")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# --- 1. Select Target q-point ---
target_q = np.array([0.2, 0, 0])  # Your desired q-point

# Find the closest q-point index from your q_coords array
distances = np.linalg.norm(q_coords - target_q, axis=1)
target_q_idx = np.argmin(distances)
actual_q = q_coords[target_q_idx]

print(f"Target q-point: {target_q}")
print(f"Closest match found at index {target_q_idx}: {actual_q}")

# --- 2. Extract Frequencies for this q-point ---
# Since w_bare is typically flattened as (Nq_irr * num_bands), 
# we extract only the frequencies belonging to the selected q-point index.
# Assuming w_bare is structured such that w_bare[q_idx + band_idx * Nq_irr] 
# or similar mapping from your previous loop: q_irr_idx = i % Nq_irr

q_point_frequencies = []
for i in range(num_total_modes):
    if (i % Nq_irr) == target_q_idx:
        q_point_frequencies.append(w_bare[i])

# --- 3. Parameter Setup ---
delta_val = 0.0
gamma_val = 0.1
dw_factor = gamma_val / 10.0

plt.figure(figsize=(10, 6))

# --- 4. Plot Spectral Densities for the selected q-point ---
print(f"\n--- Plotting {len(q_point_frequencies)} bands for q={actual_q} ---")

for b_idx, w_qs in enumerate(q_point_frequencies):
    # Skip acoustic modes at Gamma if they are near zero
    if w_qs < 0.1:
        continue
    
    # Range for plotting: Peak center +/- 10 * Gamma
    w_plot = np.linspace(w_qs + delta_val - 10 * gamma_val, 
                         w_qs + delta_val + 10 * gamma_val, 500)
    
    # Lorentzian Spectral Density
    denom = (w_plot - w_qs - delta_val)**2 + gamma_val**2
    b_w = (1.0 / np.pi) * (gamma_val / denom)
    
    # Plotting
    line, = plt.plot(w_plot, b_w, label=f'Band {b_idx+1} ({w_qs:.2f} THz)')
    plt.axvline(x=w_qs, color=line.get_color(), linestyle='--', alpha=0.3)

# --- 5. Plot Formatting ---
plt.title(f"Phonon Spectral Densities $b(\\omega)$ at q={actual_q}\n"
          f"($\\Delta$={delta_val}, $\\Gamma$={gamma_val})")
plt.xlabel("Frequency (THz)")
plt.ylabel("Spectral Density $b(\omega)$")
plt.legend(fontsize='x-small', ncol=2, loc='upper right')
plt.grid(True, alpha=0.2)
plt.tight_layout()
plt.show()

# --- 6. Normalization Check for these specific modes ---
print(f"{'Band':<8} | {'Frequency':<12} | {'Integral':<10}")
print("-" * 40)
for b_idx, w_qs in enumerate(q_point_frequencies):
    if w_qs < 0.001: continue
    
    w_min = max(0.001, w_qs + delta_val - 10 * gamma_val)
    w_max = w_qs + delta_val + 10 * gamma_val
    w_grid = np.arange(w_min, w_max, dw_factor)
    
    denom_int = (w_grid - w_qs - delta_val)**2 + gamma_val**2
    b_w_int = (1.0 / np.pi) * (gamma_val / denom_int)
    integral = np.sum(b_w_int) * dw_factor
    
    print(f"{b_idx+1:<8} | {w_qs:<12.4f} | {integral:<10.4f}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# --- 1. Reshape harmonic frequencies ---
# w_harmonic shape: (Nq_irr, num_bands)
w_harmonic = w_bare.reshape(num_bands, Nq_irr).T 

# --- 2. DOS Calculation Function ---
def calculate_harmonic_dos(frequencies, weights, energy_grid, sigma=0.05):
    """
    frequencies: (Nq_irr, num_bands)
    weights: (Nq_irr,) - Multiplicity of each q-point
    energy_grid: Frequency axis for the DOS plot
    sigma: Gaussian broadening width (THz)
    """
    dos = np.zeros_like(energy_grid)
    for i in range(len(weights)):
        for j in range(frequencies.shape[1]):
            # Add Gaussian contribution for each mode: weight * G(omega - w_ij)
            diff = energy_grid - frequencies[i, j]
            dos += weights[i] * np.exp(-0.5 * (diff / sigma)**2) / (sigma * np.sqrt(2 * np.pi))
    
    # Normalize by total number of q-points in full BZ
    return dos / np.sum(weights)

# --- 3. Setup Grid and Compute ---
# Create a frequency axis from 0 to slightly above max frequency
w_max = np.max(w_harmonic) * 1.05
energy_axis = np.linspace(0, w_max, 1000)

# Calculate DOS (You can adjust sigma for smoothness)
dos_harmonic = calculate_harmonic_dos(w_harmonic, degeneracies, energy_axis, sigma=0.04)

# --- 4. Plotting ---
plt.figure(figsize=(9, 6))
plt.plot(energy_axis, dos_harmonic, color='blue', linewidth=2, label='Harmonic DOS')
plt.fill_between(energy_axis, 0, dos_harmonic, color='blue', alpha=0.2)

plt.title('Harmonic Phonon Density of States (DOS)', fontsize=15)
plt.xlabel('Frequency (THz)', fontsize=12)
plt.ylabel('Density of States (States/THz)', fontsize=12)
plt.xlim(0, w_max)
plt.ylim(0, np.max(dos_harmonic) * 1.1)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.tight_layout()
plt.show()

print(f"Harmonic DOS plot complete.")
print(f"Frequency range: 0.0 to {np.max(w_harmonic):.3f} THz")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# --- 1. Parameters from your Cv calculation ---
gamma_qpa = 0.1  # Fixed Gamma (THz)
delta_qpa = 0.0  # Fixed Delta (THz)

# --- 2. Lorentzian Function for QPA ---
def lorentzian(w, w0, delta, gamma):
    """Lorentzian Spectral Density b_qs(w)"""
    denom = (w - (w0 + delta))**2 + gamma**2
    return (1.0 / np.pi) * (gamma / denom)

# Use the same energy axis as the harmonic DOS
energy_axis = np.linspace(0, np.max(w_harmonic) * 1.1, 1000)
dos_qpa = np.zeros_like(energy_axis)

print(f"Calculating QPA DOS with Gamma = {gamma_qpa} THz...")

# --- 3. Accumulate Lorentzian for each mode ---
# w_harmonic shape: (Nq_irr, num_bands)
for i in range(Nq_irr):
    weight = degeneracies[i]
    for j in range(num_bands):
        w_qs = w_harmonic[i, j]
        # Skip unstable modes if any
        if w_qs < 0: continue
        
        # Add Lorentzian contribution weighted by q-point degeneracy
        dos_qpa += weight * lorentzian(energy_axis, w_qs, delta_qpa, gamma_qpa)

# Normalize by total q-points (Nq_total)
dos_qpa /= Nq_total

# --- 4. Plotting Comparison ---
plt.figure(figsize=(10, 6))

# Plot Harmonic DOS (for reference)
plt.plot(energy_axis, dos_harmonic, 'k--', label='Harmonic DOS (Gaussian)', alpha=0.5)

# Plot QPA DOS
plt.plot(energy_axis, dos_qpa, color='green', linewidth=2, label=f'QPA DOS (Gamma={gamma_qpa})')
plt.fill_between(energy_axis, 0, dos_qpa, color='green', alpha=0.15)

plt.title(f'Phonon DOS: Harmonic vs. QPA (Fixed $\Gamma$)', fontsize=15)
plt.xlabel('Frequency (THz)', fontsize=12)
plt.ylabel('Density of States', fontsize=12)
plt.xlim(0, np.max(energy_axis))
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()

plt.tight_layout()
plt.show()

print("QPA DOS plotting complete.")
```
