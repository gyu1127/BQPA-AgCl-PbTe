```python
from ase.io import read
import numpy as np

# 1. Load the POSCAR
# This gives you an Atoms object which knows positions, masses, and cell
atoms = read('POSCAR')
n_atoms = len(atoms)
masses = atoms.get_masses()  # in amu
cell = atoms.get_cell()      # lattice vectors
positions = atoms.get_positions() # Cartesian coordinates
sc_dims = np.array([4, 4, 4])

print(f"Loaded {n_atoms} atoms.")
print(f"{atoms}")
```

```python
import numpy as np
from scipy.interpolate import interp1d

# --- 1. Load BTE.omega and Reorder Indices ---
# Initial data shape from file: (Nq, num_bands) -> e.g., (15625, 6)
# The file rows are organized as: q1(b1, b2...b6), q2(b1, b2...b6)
data_omega = np.loadtxt('BTE.omega') 
Nq = data_omega.shape[0]
num_bands = data_omega.shape[1]

# IMPORTANT:
# 1. data_omega.T transposes the matrix to (num_bands, Nq).
# 2. .flatten() then creates a 1D array in the order: (b1_q1, b1_q2... b1_qNq, b2_q1...)
# 3. This matches the block sequence in 'BTE.selfeng3ph_spectrum'.
# 4. Divide by 2*pi to convert rad/ps -> THz.
w_bare = (data_omega.T).flatten() / (2 * np.pi)

num_total_modes = len(w_bare)
print(f"Total modes: {num_total_modes} from BTE.omega")
print(f"Detected Nq (q-points): {Nq}, Bands: {num_bands} from BTE.omega")

# --- 2. Load and Parse BTE.selfeng3ph_spectrum ---
# We use genfromtxt to handle scientific notation 
spec_data = np.genfromtxt('BTE.selfeng3ph_spectrum')

# Calculate the number of frequency grid points per mode block
pts_per_mode = len(spec_data) // num_total_modes

d_interps = []
g_interps = []

print(f"Parsing {num_total_modes} mode blocks in BTE.selfeng3ph_spectrum...")

for i in range(num_total_modes):
    # Each block corresponds to one mode (one specific band and q-point)
    start = i * pts_per_mode
    end = (i + 1) * pts_per_mode
    mode_block = spec_data[start:end]
    
    w_grid = mode_block[:, 0] # Frequency grid (THz)
    d_grid = mode_block[:, 1] / (2*np.pi) # Real part (Delta) rad/ps -> THz
    g_grid = mode_block[:, 2] / (2*np.pi) # Imaginary part (Gamma) rad/ps -> THz
    
    # Store interpolation functions for each mode
    d_interps.append(interp1d(w_grid, d_grid, kind='linear', fill_value="extrapolate"))
    g_interps.append(interp1d(w_grid, g_grid, kind='linear', fill_value="extrapolate"))

print(f"Successfully loaded {len(d_interps)} interpolation blocks from BTE.selfeng3ph_spectrum.")
```

```python
import numpy as np
from scipy.interpolate import interp1d

# --- 1. Load BTE.qpoints to get degeneracies ---
q_data = np.genfromtxt('BTE.qpoints')
# Column 2 (index 2) is the degeneracy (multiplicity)
degeneracies = q_data[:, 2] 
Nq_irr = len(degeneracies) # the number of qpoints in irreducible wedge
Nq_total = np.sum(degeneracies) # the number of qpoints in whole BZ
q_coords = q_data[:, 3:6]

print(f"Number of irreducible q-points: {Nq_irr}")
print(f"Total q-points in full BZ (Sum of weights): {Nq_total}")
```

```python
import numpy as np

# --- Constants ---
kB = 8.617333262e-5       # Boltzmann constant in eV/K
THz_to_eV = 0.004135667   # Conversion factor (h * 1e12) to eV
T = 100.0                 # Temperature in Kelvin

total_cv_harmonic = 0.0
frequencies_harmonic = []
cv_contributions_harmonic = []
processed_modes_count = 0

for i in range(num_total_modes):
    # 1. Map global index to irreducible q-point index
    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]
    
    # 2. Get the bare frequency (Harmonic frequency)
    w_qs = w_bare[i]
    
    # 3. Handle acoustic modes at Gamma or negative frequencies
    if w_qs <= 1e-6:
        mode_cv = 0.0
    else:
        # 4. Calculate the Harmonic Heat Capacity Kernel
        # x = (hbar * omega) / (kB * T)
        x = (w_qs * THz_to_eV) / (kB * T)
        
        # Avoid numerical overflow for very large x (low T or high frequency)
        if x < 50:
            exp_x = np.exp(x)
            mode_cv = kB * (x**2) * exp_x / (exp_x - 1)**2
        else:
            mode_cv = 0.0
            
    # 5. Multiply the mode's contribution by its degeneracy weight
    total_cv_harmonic += mode_cv * weight
    processed_modes_count += 1

    # Store results for analysis
    frequencies_harmonic.append(w_qs)
    cv_contributions_harmonic.append(mode_cv)

# --- 6. Final Normalization and Unit Conversion ---
# Normalize by the total number of q-points in the full Brillouin Zone
cv_per_cell = total_cv_harmonic / Nq_total

# Convert from eV/K to J/K/mol
# (1 eV = 1.60218e-19 J, 1 mol = 6.02214e23 formula units)
cv_j_mol = cv_per_cell * 1.60218e-19 * 6.02214e23

print(f"Modes processed: {processed_modes_count}")
print(f"Harmonic Cv (J/K/mol): {cv_j_mol:.4f}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# --- 1. Reshape harmonic frequencies ---
# w_harmonic shape: (Nq_irr, num_bands)
w_harmonic = w_bare.reshape(num_bands, Nq_irr).T 

# --- 2. DOS Calculation Function ---
def calculate_harmonic_dos(frequencies, weights, energy_grid, sigma=0.05):
    """
    frequencies: (Nq_irr, num_bands)
    weights: (Nq_irr,) - Multiplicity of each q-point
    energy_grid: Frequency axis for the DOS plot
    sigma: Gaussian broadening width (THz)
    """
    dos = np.zeros_like(energy_grid)
    for i in range(len(weights)):
        for j in range(frequencies.shape[1]):
            # Add Gaussian contribution for each mode: weight * G(omega - w_ij)
            diff = energy_grid - frequencies[i, j]
            dos += weights[i] * np.exp(-0.5 * (diff / sigma)**2) / (sigma * np.sqrt(2 * np.pi))
    
    # Normalize by total number of q-points in full BZ
    return dos / np.sum(weights)

# --- 3. Setup Grid and Compute ---
# Create a frequency axis from 0 to slightly above max frequency
w_max = np.max(w_harmonic) * 1.05
energy_axis = np.linspace(0, w_max, 1000)

# Calculate DOS (You can adjust sigma for smoothness)
dos_harmonic = calculate_harmonic_dos(w_harmonic, degeneracies, energy_axis, sigma=0.04)

# --- 4. Plotting ---
plt.figure(figsize=(9, 6))
plt.plot(energy_axis, dos_harmonic, color='blue', linewidth=2, label='Harmonic DOS')
plt.fill_between(energy_axis, 0, dos_harmonic, color='blue', alpha=0.2)

plt.title('Harmonic Phonon Density of States (DOS)', fontsize=15)
plt.xlabel('Frequency (THz)', fontsize=12)
plt.ylabel('Density of States (States/THz)', fontsize=12)
plt.xlim(0, w_max)
plt.ylim(0, np.max(dos_harmonic) * 1.1)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.tight_layout()
plt.show()

print(f"Harmonic DOS plot complete.")
print(f"Frequency range: 0.0 to {np.max(w_harmonic):.3f} THz")
```
