```python
from ase.io import read
import numpy as np

# 1. Load the POSCAR
# This gives you an Atoms object which knows positions, masses, and cell
atoms = read('POSCAR')
n_atoms = len(atoms)
masses = atoms.get_masses()  # in amu
cell = atoms.get_cell()      # lattice vectors
positions = atoms.get_positions() # Cartesian coordinates
sc_dims = np.array([4, 4, 4])

print(f"Loaded {n_atoms} atoms.")
print(f"{atoms}")
```

```python
import numpy as np
from scipy.interpolate import interp1d

# --- 1. Load BTE.omega (Assumed Band-major after flattening) ---
# data_omega shape: (Nq_irr, num_bands) -> e.g., (455, 6)
data_omega = np.loadtxt('BTE.omega') 
Nq = data_omega.shape[0]
num_bands = data_omega.shape[1]

# Replicate your current logic: Band 1 (all q), then Band 2 (all q)...
# Dividing by 2*pi to convert from rad/ps to THz
w_bare = (data_omega.T).flatten() / (2 * np.pi)
num_total_modes = len(w_bare)

print(f"Total modes: {num_total_modes} from BTE.omega")
print(f"Detected Nq (q-points): {Nq}, Bands: {num_bands}")

# --- 2. Load and Parse BTE.selfeng3ph_spectrum (q-major order) ---
# Each block in this file follows: q1_b1, q1_b2... q1_b6, q2_b1...
spec_data = np.genfromtxt('BTE.selfeng3ph_spectrum')

# Calculate the number of frequency grid points per mode block
pts_per_mode = len(spec_data) // num_total_modes

# Initialize lists to store interpolation functions in Band-major order
d_interps = [None] * num_total_modes
g_interps = [None] * num_total_modes

print(f"Parsing blocks and re-mapping q-major file to Band-major w_bare...")

for i in range(num_total_modes):
    # 'i' represents the sequence in the FILE (q-major)
    # q_idx changes slowly, b_idx changes fast
    q_idx = i // num_bands
    b_idx = i % num_bands
    
    # Calculate the 'target_mode_idx' to match the Band-major 'w_bare'
    # In Band-major: b_idx changes slowly, q_idx changes fast
    target_mode_idx = b_idx * Nq + q_idx
    
    # Extract the data block for the current mode
    start = i * pts_per_mode
    end = (i + 1) * pts_per_mode
    mode_block = spec_data[start:end]
    
    w_grid = mode_block[:, 0] # Frequency grid (THz)
    # Convert Delta and Gamma from rad/ps to THz
    d_grid = mode_block[:, 1] / (2 * np.pi) 
    g_grid = mode_block[:, 2] / (2 * np.pi)
    
    # Map the interpolation function to the correct index for w_bare[target_mode_idx]
    d_interps[target_mode_idx] = interp1d(w_grid, d_grid, kind='linear', fill_value="extrapolate")
    g_interps[target_mode_idx] = interp1d(w_grid, g_grid, kind='linear', fill_value="extrapolate")

print("Successfully loaded and re-ordered all interpolation blocks!")
```

```python
# --- 3. Verification Test ---
# With the transpose, index 'i' now follows (Band_index * Nq + q_index).
# Band 2, first q-point (q=1) is located at index i = 1 * Nq + 0 = Nq.
band = 5
q = 1

band_idx = band - 1
q_idx = q - 1

test_idx = band_idx * Nq + q_idx
if test_idx < num_total_modes:
    test_w = w_bare[test_idx]
    print(f"\nVerification at Index {test_idx} (Band={band}, q={q}):")
    print(f"Bare Frequency (converted to THz): {test_w:.4f}")
    print(f"Delta at bare freq: {d_interps[test_idx](test_w)*(2*np.pi):.6f}")
    print(f"Gamma at bare freq: {g_interps[test_idx](test_w)*(2*np.pi):.6f}")
```

```python
import numpy as np
from scipy.interpolate import interp1d

# --- 1. Load BTE.qpoints to get degeneracies ---
q_data = np.genfromtxt('BTE.qpoints')
# Column 2 (index 2) is the degeneracy (multiplicity)
degeneracies = q_data[:, 2] 
Nq_irr = len(degeneracies) # the number of qpoints in irreducible wedge
Nq_total = np.sum(degeneracies) # the number of qpoints in whole BZ
q_coords = q_data[:, 3:6]

print(f"Number of irreducible q-points: {Nq_irr}")
print(f"Total q-points in full BZ (Sum of weights): {Nq_total}")
```

```python
for i in range(num_total_modes):
    w_qs = w_bare[i]
    #if w_qs < 1e-5: 
    #    continue

    q_irr_idx = i % Nq_irr
    band_irr_idx = i // Nq_irr
    weight = degeneracies[q_irr_idx]
    d_val = d_interps[i](w_qs)
    g_val = abs(g_interps[i](w_qs))
    q_coords_test = q_coords[q_irr_idx]

    print(f"i = {i}, q_irr = {q_irr_idx+1}, band_irr = {band_irr_idx+1}, weight = {weight}, w_ha = {w_qs:.4f} THz, {w_qs*(2 * np.pi):.4e} rad/ps, delta = {d_val:.4f}, gamma = {g_val:.4f}, {q_coords_test}")

# i: (b1,q1), (b1, q2), (b1, q3), ...
```

```python
total_cv_weighted = 0.0
frequencies_qpa_d0 = []
cv_contributions_qpa_d0 = []

kB = 8.617333262e-5      # Boltzmann constant in eV/K
THz_to_eV = 0.004135667  # Conversion factor (h * 1e12)
T = 150.0                # Temperature in Kelvin
integrated_modes_count = 0

for i in range(num_total_modes):
    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]
    
    w_qs = w_bare[i]
    #if w_qs < 1e-10: continue
    
    # 1. Get Delta and Gamma for this specific irreducible mode
    #d_val = d_interps[i](w_qs)
    d_val = 0
    g_val = abs(g_interps[i](w_qs))
    
    # 2. Strict BQPA Lorentzian Integration
    g_floor = max(g_val, 1e-4) # prevent gamma=0
    dw = g_floor / 10.0 # mode-adative
    w_min = max(0.001, w_qs + d_val - 10 * g_floor)
    w_max = w_qs + d_val + 10 * g_floor
    local_grid = np.arange(w_min, w_max, dw)
    
    mode_integral = 0.0
    for w in local_grid:
        # Spectral Density b_qs(w)
        denom = (w - w_qs - d_val)**2 + g_val**2
        b_w = (1/np.pi) * (g_val / denom)
        
        # Heat Capacity Kernel
        x = (w * THz_to_eV) / (kB * T)
        if x < 50:
            cv_kernel = kB * (x**2) * np.exp(x) / (np.exp(x) - 1)**2
            mode_integral += b_w * cv_kernel * dw
    
    # Multiply the mode's contribution by its degeneracy weight
    total_cv_weighted += mode_integral * weight
    integrated_modes_count += 1

    frequencies_qpa_d0.append(w_qs)
    cv_contributions_qpa_d0.append(mode_integral)
    
# --- 3. Final Normalization by total BZ points ---
cv_per_cell = total_cv_weighted / Nq_total
cv_j_mol = cv_per_cell * 1.60218e-19 * 6.02214e23
print(f"Modes used (integrated): {integrated_modes_count}")
print(f"Cv of unitcell(mol of unitcell): {cv_j_mol:.4f} J/K/mol")
```

```python
total_cv_weighted = 0.0
frequencies_qpa = []
cv_contributions_qpa = []

kB = 8.617333262e-5      # Boltzmann constant in eV/K
THz_to_eV = 0.004135667  # Conversion factor (h * 1e12)
T = 150.0                # Temperature in Kelvin

for i in range(num_total_modes):
    # Mapping the global index 'i' back to the irreducible q-point index
    # Since w_bare is (num_bands * Nq_irr), the q-index repeats every Nq_irr
    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]
    
    w_qs = w_bare[i]
    #if w_qs < 1e-10: continue
    
    # 1. Get Delta and Gamma for this specific irreducible mode
    d_val = d_interps[i](w_qs)
    #d_val = 0
    g_val = abs(g_interps[i](w_qs))
    
    # 2. Strict BQPA Lorentzian Integration
    g_floor = max(g_val, 1e-4) # prevent gamma=0
    dw = g_floor / 10.0 # mode-adative
    w_min = max(0.001, w_qs + d_val - 10 * g_floor)
    w_max = w_qs + d_val + 10 * g_floor
    local_grid = np.arange(w_min, w_max, dw)
    
    mode_integral = 0.0
    for w in local_grid:
        # Spectral Density b_qs(w)
        denom = (w - w_qs - d_val)**2 + g_val**2
        b_w = (1/np.pi) * (g_val / denom)
        
        # Heat Capacity Kernel
        x = (w * THz_to_eV) / (kB * T)
        if x < 50:
            cv_kernel = kB * (x**2) * np.exp(x) / (np.exp(x) - 1)**2
            mode_integral += b_w * cv_kernel * dw
    
    # Multiply the mode's contribution by its degeneracy weight
    total_cv_weighted += mode_integral * weight

    frequencies_qpa.append(w_qs)
    cv_contributions_qpa.append(mode_integral)
    
# --- 3. Final Normalization by total BZ points ---
cv_per_cell = total_cv_weighted / Nq_total
cv_j_mol = cv_per_cell * 1.60218e-19 * 6.02214e23

print(f"Cv of unitcell(mol of unitcell): {cv_j_mol:.4f} J/K/mol")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# --- 1. Reshape harmonic frequencies ---
# w_harmonic shape: (Nq_irr, num_bands)
w_harmonic = w_bare.reshape(num_bands, Nq_irr).T 

# --- 2. DOS Calculation Function ---
def calculate_harmonic_dos(frequencies, weights, energy_grid, sigma=0.05):
    """
    frequencies: (Nq_irr, num_bands)
    weights: (Nq_irr,) - Multiplicity of each q-point
    energy_grid: Frequency axis for the DOS plot
    sigma: Gaussian broadening width (THz)
    """
    dos = np.zeros_like(energy_grid)
    for i in range(len(weights)):
        for j in range(frequencies.shape[1]):
            # Add Gaussian contribution for each mode: weight * G(omega - w_ij)
            diff = energy_grid - frequencies[i, j]
            dos += weights[i] * np.exp(-0.5 * (diff / sigma)**2) / (sigma * np.sqrt(2 * np.pi))
    
    # Normalize by total number of q-points in full BZ
    return dos / np.sum(weights)

# --- 3. Setup Grid and Compute ---
# Create a frequency axis from 0 to slightly above max frequency
w_max = np.max(w_harmonic) * 1.05
energy_axis = np.linspace(0, w_max, 1000)

# Calculate DOS (You can adjust sigma for smoothness)
dos_harmonic = calculate_harmonic_dos(w_harmonic, degeneracies, energy_axis, sigma=0.04)

# --- 4. Plotting ---
plt.figure(figsize=(9, 6))
plt.plot(energy_axis, dos_harmonic, color='blue', linewidth=2, label='Harmonic DOS')
plt.fill_between(energy_axis, 0, dos_harmonic, color='blue', alpha=0.2)

plt.title('Harmonic Phonon Density of States (DOS)', fontsize=15)
plt.xlabel('Frequency (THz)', fontsize=12)
plt.ylabel('Density of States (States/THz)', fontsize=12)
plt.xlim(0, w_max)
plt.ylim(0, np.max(dos_harmonic) * 1.1)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.tight_layout()
plt.show()

print(f"Harmonic DOS plot complete.")
print(f"Frequency range: 0.0 to {np.max(w_harmonic):.3f} THz")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# --- 1. Prepare DOS Grid ---
energy_axis = np.linspace(0, np.max(w_harmonic) * 1.1, 1000)
dos_qpa_varied = np.zeros_like(energy_axis)

print("Calculating QPA DOS using mode-specific Self-Energy...")

# --- 2. Accumulate Lorentzians with mode-specific parameters ---
for i in range(num_total_modes):
    # Mapping back to the irreducible weight
    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]
    
    w_qs = w_bare[i]
    
    # 1. Get Delta and Gamma at the harmonic frequency (As per your Cv code)
    d_val = d_interps[i](w_qs)
    g_val = abs(g_interps[i](w_qs))
    
    # 2. Define the Lorentzian for this specific mode
    # Peak is at (w_qs + d_val), width is g_val
    denom = (energy_axis - (w_qs + d_val))**2 + g_val**2
    # Prevent division by zero if g_val is 0
    if g_val > 1e-6:
        b_w = (1.0 / np.pi) * (g_val / denom)
    else:
        # Fallback to a very narrow peak if Gamma is near zero
        b_w = np.exp(-0.5 * ((energy_axis - (w_qs + d_val)) / 0.01)**2) / (0.01 * np.sqrt(2 * np.pi))
        
    # 3. Add to total DOS with degeneracy weight
    dos_qpa_varied += weight * b_w

# --- 3. Final Normalization ---
dos_qpa_varied /= Nq_total

# --- 4. Plotting Comparison ---
plt.figure(figsize=(10, 6))

# Harmonic for reference
plt.plot(energy_axis, dos_harmonic, 'k--', label='Harmonic DOS', alpha=0.4)

# QPA with varied parameters
plt.plot(energy_axis, dos_qpa_varied, color='green', linewidth=2, label='QPA DOS (Mode-specific $\Sigma$)')
plt.fill_between(energy_axis, 0, dos_qpa_varied, color='green', alpha=0.15)

plt.title('Phonon DOS: Harmonic vs. QPA (Mode-specific)', fontsize=15)
plt.xlabel('Frequency (THz)', fontsize=12)
plt.ylabel('Density of States', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()

plt.tight_layout()
plt.show()

print("QPA (Mode-specific) DOS plotting complete.")
```

```python
import numpy as np

total_cv_full = 0.0
frequencies_bqpa = []
cv_contributions_bqpa = []

# Define global frequency bounds from the loaded spectrum file
global_w_min = spec_data[:, 0].min()
global_w_max = spec_data[:, 0].max()

print(f"Starting Full Frequency-Dependent BQPA integration...")

for i in range(num_total_modes):
    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]
    w_qs = w_bare[i]
    
    if w_qs < 1e-10: continue

    # Local step size based on broadening at resonance
    g_ref = abs(g_interps[i](w_qs))
    dw = max(g_ref, 1e-4) / 10.0
    
    # Integration grid for this mode
    w_grid = np.arange(global_w_min, global_w_max, dw)
    
    # Evaluate frequency-dependent Delta and Gamma
    d_ws = d_interps[i](w_grid)
    g_ws = np.abs(g_interps[i](w_grid))
    
    # Spectral Density b_qs(w)
    denoms = (w_grid - w_qs - d_ws)**2 + g_ws**2
    b_ws = (1/np.pi) * (g_ws / denoms)
    
    # Heat Capacity Kernel with Singularity Fix
    x = (w_grid * THz_to_eV) / (kB * T)
    cv_kernels = np.zeros_like(x)
    
    # Handle x=0 limit (kB) and general case
    m_zero = (x < 1e-7)
    m_finite = (x >= 1e-7) & (x < 50)
    
    cv_kernels[m_zero] = kB
    cv_kernels[m_finite] = kB * (x[m_finite]**2 * np.exp(x[m_finite])) / (np.exp(x[m_finite]) - 1)**2
    
    # Mode contribution (Area under b_w * Cv_kernel)
    mode_integral = np.sum(b_ws * cv_kernels) * dw
    total_cv_weighted = mode_integral * weight
    
    total_cv_full += total_cv_weighted
    frequencies_bqpa.append(w_qs)
    cv_contributions_bqpa.append(mode_integral)

# Final Normalization
cv_per_cell = total_cv_full / Nq_total
cv_j_mol = cv_per_cell * 1.60218e-19 * 6.02214e23

print(f"Full Frequency-Dependent Cv: {cv_j_mol:.4f} J/K/mol")
```

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_normalized_modes_simple(file_path):
    with open(file_path, 'r') as f:
        content = f.read().strip().split('\n')
    
    modes_list = []
    current_mode = {}
    temp_grid_density = []

    for line in content:
        parts = line.split()
        if len(parts) == 1:
            if current_mode:
                half = len(temp_grid_density) // 2
                current_mode['x'] = np.array(temp_grid_density[:half]) / (2 * np.pi)
                current_mode['y'] = np.array(temp_grid_density[half:])
                modes_list.append(current_mode)
            current_mode = {'center': float(parts[0]) / (2 * np.pi)}
            temp_grid_density = []
        else:
            temp_grid_density.extend([float(p) for p in parts])

    if current_mode:
        half = len(temp_grid_density) // 2
        current_mode['x'] = np.array(temp_grid_density[:half]) / (2 * np.pi)
        current_mode['y'] = np.array(temp_grid_density[half:])
        modes_list.append(current_mode)

    plt.figure(figsize=(10, 6))

    for idx, m in enumerate(modes_list):
        norm_factor = np.trapz(m['y'], m['x'])
        y_norm = m['y'] / norm_factor
        plt.plot(m['x'], y_norm/ (2 * np.pi), label=f'Mode {idx+1}')

    plt.xlabel('Frequency (THz)')
    plt.ylabel('Normalized Spectral Density')
    plt.title('Normalized Vibrational Modes')
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()


plot_normalized_modes_simple('bspec_omega_5.dat')
```

```python
import matplotlib.pyplot as plt
import numpy as np

# --- 1. Setup Energy Grid ---
# Use the global bounds from your spectrum data
energy_axis = np.linspace(global_w_min, global_w_max, 1000)
dos_bqpa_full = np.zeros_like(energy_axis)

print("Calculating Full BQPA DOS (Frequency-dependent Spectral Density)...")

# --- 2. Accumulate Full Spectral Density for each mode ---
for i in range(num_total_modes):
    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]
    w_qs = w_bare[i]
    
    # 1. Evaluate Delta(w) and Gamma(w) on the entire energy axis
    d_ws = d_interps[i](energy_axis)
    g_ws = np.abs(g_interps[i](energy_axis))
    
    # 2. Calculate the Full Spectral Density b_qs(w)
    # This reflects the true shape of the phonon state in an anharmonic potential
    denoms = (energy_axis - w_qs - d_ws)**2 + g_ws**2
    
    # Avoid division by zero for very stable modes
    g_ws_safe = np.where(g_ws < 1e-6, 1e-6, g_ws)
    b_ws = (1.0 / np.pi) * (g_ws_safe / denoms)
    
    # 3. Add to total DOS with degeneracy weight
    dos_bqpa_full += weight * b_ws

# --- 3. Final Normalization ---
dos_bqpa_full /= Nq_total

# --- 4. Plotting Comparison: Harmonic vs QPA vs BQPA ---
plt.figure(figsize=(12, 7))

# 1. Harmonic DOS (Reference)
plt.plot(energy_axis, dos_harmonic, 'k--', label='Harmonic DOS', alpha=0.3)

# 2. QPA DOS (From previous step - if you have the variable)
#if 'dos_qpa_varied' in locals():
#    plt.plot(energy_axis, dos_qpa_varied, color='green', label='QPA DOS (Single-point $\Sigma$)', alpha=0.5)

# 3. Full BQPA DOS
plt.plot(energy_axis, dos_bqpa_full, color='purple', linewidth=2.5, label='Full BQPA DOS (Freq-dependent $\Sigma$)')
plt.fill_between(energy_axis, 0, dos_bqpa_full, color='purple', alpha=0.15)

plt.title('Phonon DOS: Full BQPA Comparison', fontsize=16)
plt.xlabel('Frequency (THz)', fontsize=13)
plt.ylabel('Density of States', fontsize=13)
plt.xlim(0, global_w_max)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()

plt.tight_layout()
plt.show()
```
