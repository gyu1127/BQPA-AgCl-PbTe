```python
from ase.io import read
import numpy as np

# 1. Load the POSCAR
# This gives you an Atoms object which knows positions, masses, and cell
atoms = read('POSCAR')
n_atoms = len(atoms)
masses = atoms.get_masses()  # in amu
cell = atoms.get_cell()      # lattice vectors
positions = atoms.get_positions() # Cartesian coordinates
sc_dims = np.array([4, 4, 4])

print(f"Loaded {n_atoms} atoms.")
print(f"{atoms}")
```

```python
import numpy as np
from scipy.interpolate import interp1d

# --- 1. Load BTE.omega (Assumed Band-major after flattening) ---
# data_omega shape: (Nq_irr, num_bands) -> e.g., (455, 6)
data_omega = np.loadtxt('BTE.omega') 
Nq = data_omega.shape[0]
num_bands = data_omega.shape[1]

# Replicate your current logic: Band 1 (all q), then Band 2 (all q)...
# Dividing by 2*pi to convert from rad/ps to THz
w_bare = (data_omega.T).flatten() / (2 * np.pi)
num_total_modes = len(w_bare)

print(f"Total modes: {num_total_modes} from BTE.omega")
print(f"Detected Nq (q-points): {Nq}, Bands: {num_bands}")

# --- 2. Load and Parse BTE.selfeng3ph_spectrum (q-major order) ---
# Each block in this file follows: q1_b1, q1_b2... q1_b6, q2_b1...
spec_data = np.genfromtxt('BTE.selfeng3ph_spectrum')

# Calculate the number of frequency grid points per mode block
pts_per_mode = len(spec_data) // num_total_modes

# Initialize lists to store interpolation functions in Band-major order
d_interps = [None] * num_total_modes
g_interps = [None] * num_total_modes

print(f"Parsing blocks and re-mapping q-major file to Band-major w_bare...")

for i in range(num_total_modes):
    # 'i' represents the sequence in the FILE (q-major)
    # q_idx changes slowly, b_idx changes fast
    q_idx = i // num_bands
    b_idx = i % num_bands
    
    # Calculate the 'target_mode_idx' to match the Band-major 'w_bare'
    # In Band-major: b_idx changes slowly, q_idx changes fast
    target_mode_idx = b_idx * Nq + q_idx
    
    # Extract the data block for the current mode
    start = i * pts_per_mode
    end = (i + 1) * pts_per_mode
    mode_block = spec_data[start:end]
    
    w_grid = mode_block[:, 0] # Frequency grid (THz)
    # Convert Delta and Gamma from rad/ps to THz
    d_grid = mode_block[:, 1] / (2 * np.pi) 
    g_grid = mode_block[:, 2] / (2 * np.pi)
    
    # Map the interpolation function to the correct index for w_bare[target_mode_idx]
    d_interps[target_mode_idx] = interp1d(w_grid, d_grid, kind='linear', fill_value="extrapolate")
    g_interps[target_mode_idx] = interp1d(w_grid, g_grid, kind='linear', fill_value="extrapolate")

print("Successfully loaded and re-ordered all interpolation blocks!")
```

```python
import numpy as np
from scipy.interpolate import interp1d

# --- 1. Load BTE.qpoints to get degeneracies ---
q_data = np.genfromtxt('BTE.qpoints')
# Column 2 (index 2) is the degeneracy (multiplicity)
degeneracies = q_data[:, 2] 
Nq_irr = len(degeneracies) # the number of qpoints in irreducible wedge
Nq_total = np.sum(degeneracies) # the number of qpoints in whole BZ
q_coords = q_data[:, 3:6]

print(f"Number of irreducible q-points: {Nq_irr}")
print(f"Total q-points in full BZ (Sum of weights): {Nq_total}")
```

```python
import numpy as np

# --- Constants ---
kB = 8.617333262e-5       # Boltzmann constant in eV/K
THz_to_eV = 0.004135667   # Conversion factor (h * 1e12) to eV
T = 300.0                 # Temperature in Kelvin

# Fixed Anharmonic Parameters
delta_fixed = 0.0         # Frequency shift (Delta)
gamma_fixed = 0.1         # Broadening (Gamma) in THz

total_cv_weighted = 0.0
frequencies_01 = []
cv_contributions_01 = []
integrated_modes_count = 0

for i in range(num_total_modes):
    # 1. Map global index to irreducible q-point index
    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]
    w_qs = w_bare[i]
    
    # Skip unstable or extremely low frequency modes
    #if w_qs < 1e-10: continue

    # 2. Setup Integration Grid for Spectral Density
    g_val = gamma_fixed
    d_val = delta_fixed
    
    # Use adaptive step size
    dw = g_val / 10.0 
    
    # Integration range: +/- 10*Gamma around the shifted peak
    w_min = max(0.001, w_qs + d_val - 100 * g_val)
    w_max = w_qs + d_val + 100 * g_val
    local_grid = np.arange(w_min, w_max, dw)
    
    mode_integral = 0.0
    norm_sum = 0.0  # Used for Spectral Density Normalization
    
    # 3. Numerical Integration
    for w in local_grid:
        # Lorentzian Spectral Density: b_qs(w)
        # Note: (1/pi) * (gamma / ( (w-w_0)^2 + gamma^2 ))
        denom = (w - w_qs - d_val)**2 + g_val**2
        b_w = (1.0 / np.pi) * (g_val / denom)
        
        # Heat Capacity Kernel
        x = (w * THz_to_eV) / (kB * T)
        
        if x < 50:
            cv_kernel = kB * (x**2) * np.exp(x) / (np.exp(x) - 1.0)**2
            
            # Summing values for the integral
            mode_integral += b_w * cv_kernel * dw
            norm_sum += b_w * dw
    
    # 4. Normalize the Spectral Density Integral
    # Even with a +/- 10*Gamma range, the Lorentzian integral is ~0.937.
    # We divide by norm_sum to ensure the total probability is exactly 1.0.
    #if norm_sum > 0:
    #    mode_integral /= norm_sum
    
    # 5. Accumulate Results
    total_cv_weighted += mode_integral * weight
    integrated_modes_count += 1

    frequencies_01.append(w_qs)
    cv_contributions_01.append(mode_integral)

# --- 6. Final Normalization and Unit Conversion ---
# Normalize by the total number of q-points in the Brillouin Zone
cv_per_cell = total_cv_weighted / Nq_total

# Convert from eV/K/cell to J/K/mol
# (1.60218e-19 J/eV * 6.02214e23 / mol)
cv_j_mol = cv_per_cell * 1.60218e-19 * 6.02214e23

print(f"Total modes integrated: {integrated_modes_count}")
print(f"BQPA Cv (Gamma=0.1, Normalized): {cv_j_mol:.4f} J/K/mol")
```

```python
total_cv_weighted = 0.0
frequencies_qpa = []
cv_contributions_qpa = []

kB = 8.617333262e-5      # Boltzmann constant in eV/K
THz_to_eV = 0.004135667  # Conversion factor (h * 1e12)
#T = 100.0                # Temperature in Kelvin

for i in range(num_total_modes):
    # Mapping the global index 'i' back to the irreducible q-point index
    # Since w_bare is (num_bands * Nq_irr), the q-index repeats every Nq_irr
    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]
    
    w_qs = w_bare[i]
    #if w_qs < 1e-10: continue
    
    # 1. Get Delta and Gamma for this specific irreducible mode
    d_val = d_interps[i](w_qs)
    #d_val = 0
    g_val = abs(g_interps[i](w_qs))
    
    # 2. Strict BQPA Lorentzian Integration
    g_floor = max(g_val, 1e-4) # prevent gamma=0
    dw = g_floor / 10.0 # mode-adative
    w_min = max(0.001, w_qs + d_val - 100 * g_floor)
    w_max = w_qs + d_val + 100 * g_floor
    local_grid = np.arange(w_min, w_max, dw)
    
    mode_integral = 0.0
    for w in local_grid:
        # Spectral Density b_qs(w)
        denom = (w - w_qs - d_val)**2 + g_val**2
        b_w = (1/np.pi) * (g_val / denom)
        
        # Heat Capacity Kernel
        x = (w * THz_to_eV) / (kB * T)
        if x < 50:
            cv_kernel = kB * (x**2) * np.exp(x) / (np.exp(x) - 1)**2
            mode_integral += b_w * cv_kernel * dw
    
    # Multiply the mode's contribution by its degeneracy weight
    total_cv_weighted += mode_integral * weight

    frequencies_qpa.append(w_qs)
    cv_contributions_qpa.append(mode_integral)
    
# --- 3. Final Normalization by total BZ points ---
cv_per_cell = total_cv_weighted / Nq_total
cv_j_mol = cv_per_cell * 1.60218e-19 * 6.02214e23

print(f"Cv of unitcell(mol of unitcell): {cv_j_mol:.4f} J/K/mol")
```

```python
import numpy as np

total_cv_full = 0.0
frequencies_bqpa = []
cv_contributions_bqpa = []


# Define global frequency bounds from the loaded spectrum file
global_w_min = spec_data[:, 0].min()
global_w_max = spec_data[:, 0].max()

print(f"Starting Full Frequency-Dependent BQPA integration...")

for i in range(num_total_modes):
    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]
    w_qs = w_bare[i]
    
    if w_qs < 1e-10: continue

    # Local step size based on broadening at resonance
    g_ref = abs(g_interps[i](w_qs))
    dw = max(g_ref, 1e-4) / 10.0
    
    # Integration grid for this mode
    w_grid = np.arange(global_w_min, global_w_max, dw)
    
    # Evaluate frequency-dependent Delta and Gamma
    d_ws = d_interps[i](w_grid)
    g_ws = np.abs(g_interps[i](w_grid))
    
    # Spectral Density b_qs(w)
    denoms = (w_grid - w_qs - d_ws)**2 + g_ws**2
    b_ws = (1/np.pi) * (g_ws / denoms)
    
    # Heat Capacity Kernel with Singularity Fix
    x = (w_grid * THz_to_eV) / (kB * T)
    cv_kernels = np.zeros_like(x)
    
    # Handle x=0 limit (kB) and general case
    m_zero = (x < 1e-7)
    m_finite = (x >= 1e-7) & (x < 50)
    
    cv_kernels[m_zero] = kB
    cv_kernels[m_finite] = kB * (x[m_finite]**2 * np.exp(x[m_finite])) / (np.exp(x[m_finite]) - 1)**2
    
    # Mode contribution (Area under b_w * Cv_kernel)
    mode_integral = np.sum(b_ws * cv_kernels) * dw
    total_cv_weighted = mode_integral * weight
    
    total_cv_full += total_cv_weighted
    frequencies_bqpa.append(w_qs)
    cv_contributions_bqpa.append(mode_integral)

# Final Normalization
cv_per_cell = total_cv_full / Nq_total
cv_j_mol = cv_per_cell * 1.60218e-19 * 6.02214e23

print(f"Full Frequency-Dependent Cv: {cv_j_mol:.4f} J/K/mol")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import AutoMinorLocator


plt.figure(figsize=(8, 8))

Unit_conv = 1.60218e-19 * 6.02214e23
# 1. Harmonic Reference Curve (Red Dashed Line)
# Combining the frequency range from all datasets for the reference line
all_freqs = np.concatenate([frequencies_01, frequencies_qpa, frequencies_bqpa])
if len(all_freqs) > 0:
    w_ref = np.linspace(min(all_freqs), max(all_freqs), 500)
    x_ref = (w_ref * THz_to_eV) / (kB * T)
    cv_harmonic = kB * (x_ref**2) * np.exp(x_ref) / (np.exp(x_ref) - 1)**2
    
    plt.plot(w_ref, cv_harmonic * Unit_conv, 
             color='SlateGrey', linestyle='--', linewidth=3, label='Harmonic', zorder=1)

# 2. QPA gamma=0.1 (Slate Blue)
plt.scatter(frequencies_01, np.array(cv_contributions_01) * Unit_conv, 
            alpha=0.7, s=15, color='#0047AB', label='Method 1', zorder=2)

# 3. QPA gamma=1.0 (Emerald Green)
plt.scatter(frequencies_qpa, np.array(cv_contributions_qpa) * Unit_conv, 
            alpha=0.7, s=15, color='#2E8B57', label='Method 2', zorder=3)

# 4. BQPA (Golden Rod)
plt.scatter(frequencies_bqpa, np.array(cv_contributions_bqpa) * Unit_conv, 
            alpha=0.7, s=15, color='#DAA520', label='Method 3', zorder=4)

# --- Formatting for Publication ---
plt.xlabel('Frequency (THz)', fontsize=22)
plt.ylabel('Mode heat capacity (J/mol$\cdot$K)', fontsize=22)

plt.xticks(fontsize=18)
plt.yticks(fontsize=18)


# Legend with professional styling
plt.legend(loc='best', fontsize=14, edgecolor='black', fancybox=False)
plt.xlim(0, 6.3)
plt.ylim(7, 8.6)
# Grid and Ticks
plt.grid(False)
plt.tick_params(axis='both', which='major', width=1.5, length=7, direction='in')
plt.minorticks_on()
plt.gca().xaxis.set_minor_locator(AutoMinorLocator(2))
plt.gca().yaxis.set_minor_locator(AutoMinorLocator(2))
plt.tick_params(axis='both', which='minor', width=1.0, length=4, direction='in')

# Spine thickness
for spine in plt.gca().spines.values():
    spine.set_linewidth(1.5)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import AutoMinorLocator

# --- 1. Setup Global Energy Axis ---
# Use a consistent range for all DOS calculations
w_max_limit = np.max(w_bare) * 1.4
energy_axis = np.linspace(0, w_max_limit, 1000)
Nq_total = np.sum(degeneracies)

# --- 2. DOS Calculation Containers ---
dos_harmonic = np.zeros_like(energy_axis)
dos_method1 = np.zeros_like(energy_axis)  # QPA Fixed
dos_method2 = np.zeros_like(energy_axis)  # QPA Varied
dos_method3 = np.zeros_like(energy_axis)  # BQPA Full

print("Calculating all DOS datasets...")

# --- 3. Compute Each Method ---

# A. Harmonic DOS (Gaussian Broadening)
sigma = 0.04
w_harmonic_flat = w_bare.reshape(-1)
for i, w_qs in enumerate(w_bare):
    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]
    diff = energy_axis - w_qs
    dos_harmonic += weight * np.exp(-0.5 * (diff / sigma)**2) / (sigma * np.sqrt(2 * np.pi))

# B. Method 1: QPA Fixed (Gamma=0.1, Delta=0)
gamma_fixed = 0.1
for i, w_qs in enumerate(w_bare):
    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]
    denom = (energy_axis - w_qs)**2 + gamma_fixed**2
    dos_method1 += weight * (1.0 / np.pi) * (gamma_fixed / denom)

# C. Method 2: QPA Mode-specific (Single-point Sigma)
for i in range(num_total_modes):
    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]
    w_qs = w_bare[i]
    d_val = d_interps[i](w_qs)
    g_val = max(abs(g_interps[i](w_qs)), 1e-4)
    denom = (energy_axis - (w_qs + d_val))**2 + g_val**2
    dos_method2 += weight * (1.0 / np.pi) * (g_val / denom)

# D. Method 3: BQPA Full (Frequency-dependent Sigma)
for i in range(num_total_modes):
    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]
    w_qs = w_bare[i]
    d_ws = d_interps[i](energy_axis)
    g_ws = np.maximum(np.abs(g_interps[i](energy_axis)), 1e-4)
    denom = (energy_axis - w_qs - d_ws)**2 + g_ws**2
    dos_method3 += weight * (1.0 / np.pi) * (g_ws / denom)

# --- 4. Final Normalization ---
dos_harmonic /= Nq_total
dos_method1 /= Nq_total
dos_method2 /= Nq_total
dos_method3 /= Nq_total

# --- 5. Integrated Plotting ---
plt.figure(figsize=(10, 8))

# Harmonic: Grey Dashed Line (The Reference)
plt.plot(energy_axis, dos_harmonic, color='#666666', linestyle='--', 
         linewidth=2.5, label='Harmonic', zorder=1)

# Method 1: Deep Royal Blue
plt.plot(energy_axis, dos_method1, color='#0047AB', linestyle='-', 
         linewidth=2.5, label='Method 1', zorder=2)
plt.fill_between(energy_axis, 0, dos_method1, color='#0047AB', alpha=0.05)

# Method 2: Emerald Green
plt.plot(energy_axis, dos_method2, color='#2E8B57', linestyle='-', 
         linewidth=2.5, label='Method 2', zorder=3)
plt.fill_between(energy_axis, 0, dos_method2, color='#2E8B57', alpha=0.05)

# Method 3: Golden Rod
plt.plot(energy_axis, dos_method3, color='#DAA520', linestyle='-', 
         linewidth=2.5, label='Method 3', zorder=4)
plt.fill_between(energy_axis, 0, dos_method3, color='#DAA520', alpha=0.1)

# --- Formatting ---
plt.xlabel('Frequency (THz)', fontsize=25)
plt.ylabel('Density of States (THz$^{-1}$)', fontsize=25)
plt.xticks(fontsize=25)
plt.yticks(fontsize=25)
plt.xlim(0, 7)
plt.ylim(0, np.max(dos_method3)*1.1)

plt.legend(loc='upper right', fontsize=18, edgecolor='black', fancybox=False)
plt.grid(False)
plt.ylim(0, 3)

# Ticks and Spines
plt.tick_params(axis='both', which='major', width=2.0, length=8, direction='in')
plt.minorticks_on()
plt.gca().xaxis.set_minor_locator(AutoMinorLocator(2))
plt.gca().yaxis.set_minor_locator(AutoMinorLocator(2))
plt.tick_params(axis='both', which='minor', width=2.0, length=4, direction='in')

for spine in plt.gca().spines.values():
    spine.set_linewidth(2.0)
    spine.set_zorder(10)

plt.tight_layout()
plt.show()

print("Combined DOS plot complete.")
```
