```python
from ase.io import read
import numpy as np

# 1. Load the POSCAR
# This gives you an Atoms object which knows positions, masses, and cell
atoms = read('POSCAR')
n_atoms = len(atoms)
masses = atoms.get_masses()  # in amu
cell = atoms.get_cell()      # lattice vectors
positions = atoms.get_positions() # Cartesian coordinates
sc_dims = np.array([4, 4, 4])

print(f"Loaded {n_atoms} atoms.")
print(f"{atoms}")
```

```python
import numpy as np
from scipy.interpolate import interp1d

# --- 1. Load BTE.omega (Assumed Band-major after flattening) ---
# data_omega shape: (Nq_irr, num_bands) -> e.g., (455, 6)
data_omega = np.loadtxt('BTE.omega') 
Nq = data_omega.shape[0]
num_bands = data_omega.shape[1]

# Replicate your current logic: Band 1 (all q), then Band 2 (all q)...
# Dividing by 2*pi to convert from rad/ps to THz
w_bare = (data_omega.T).flatten() / (2 * np.pi)
num_total_modes = len(w_bare)

print(f"Total modes: {num_total_modes} from BTE.omega")
print(f"Detected Nq (q-points): {Nq}, Bands: {num_bands}")

# --- 2. Load and Parse BTE.selfeng3ph_spectrum (q-major order) ---
# Each block in this file follows: q1_b1, q1_b2... q1_b6, q2_b1...
spec_data = np.genfromtxt('BTE.selfeng3ph_spectrum')

# Calculate the number of frequency grid points per mode block
pts_per_mode = len(spec_data) // num_total_modes

# Initialize lists to store interpolation functions in Band-major order
d_interps = [None] * num_total_modes
g_interps = [None] * num_total_modes

print(f"Parsing blocks and re-mapping q-major file to Band-major w_bare...")

for i in range(num_total_modes):
    # 'i' represents the sequence in the FILE (q-major)
    # q_idx changes slowly, b_idx changes fast
    q_idx = i // num_bands
    b_idx = i % num_bands
    
    # Calculate the 'target_mode_idx' to match the Band-major 'w_bare'
    # In Band-major: b_idx changes slowly, q_idx changes fast
    target_mode_idx = b_idx * Nq + q_idx
    
    # Extract the data block for the current mode
    start = i * pts_per_mode
    end = (i + 1) * pts_per_mode
    mode_block = spec_data[start:end]
    
    w_grid = mode_block[:, 0] # Frequency grid (THz)
    # Convert Delta and Gamma from rad/ps to THz
    d_grid = mode_block[:, 1] / (2 * np.pi) 
    g_grid = mode_block[:, 2] / (2 * np.pi)
    
    # Map the interpolation function to the correct index for w_bare[target_mode_idx]
    d_interps[target_mode_idx] = interp1d(w_grid, d_grid, kind='linear', fill_value="extrapolate")
    g_interps[target_mode_idx] = interp1d(w_grid, g_grid, kind='linear', fill_value="extrapolate")

print("Successfully loaded and re-ordered all interpolation blocks!")
```

```python
# --- 3. Verification Test ---
# With the transpose, index 'i' now follows (Band_index * Nq + q_index).
# Band 2, first q-point (q=1) is located at index i = 1 * Nq + 0 = Nq.
band = 5
q = 1

band_idx = band - 1
q_idx = q - 1

test_idx = band_idx * Nq + q_idx
if test_idx < num_total_modes:
    test_w = w_bare[test_idx]
    print(f"\nVerification at Index {test_idx} (Band={band}, q={q}):")
    print(f"Bare Frequency (converted to THz): {test_w:.4f}")
    print(f"Delta at bare freq: {d_interps[test_idx](test_w)*(2*np.pi):.6f}")
    print(f"Gamma at bare freq: {g_interps[test_idx](test_w)*(2*np.pi):.6f}")
```

```python
import numpy as np
from scipy.interpolate import interp1d

# --- 1. Load BTE.qpoints to get degeneracies ---
q_data = np.genfromtxt('BTE.qpoints')
# Column 2 (index 2) is the degeneracy (multiplicity)
degeneracies = q_data[:, 2] 
Nq_irr = len(degeneracies) # the number of qpoints in irreducible wedge
Nq_total = np.sum(degeneracies) # the number of qpoints in whole BZ
q_coords = q_data[:, 3:6]

print(f"Number of irreducible q-points: {Nq_irr}")
print(f"Total q-points in full BZ (Sum of weights): {Nq_total}")
```

```python
total_cv_weighted = 0.0
frequencies_qpa_d0 = []
cv_contributions_qpa_d0 = []

# diagnostic containers
imaginary_modes = []
spectral_loss_modes = []
tail_negative_modes = []

kB = 8.617333262e-5
THz_to_eV = 0.004135667
T = 300.0

integrated_modes_count = 0

for i in range(num_total_modes):

    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]

    w_qs = w_bare[i]

    # self-energy
    d_val = 0
    g_val = abs(g_interps[i](w_qs))

    # renormalized frequency
    w_re = w_qs + d_val

    # tail check
    if w_re - 10 * g_val < 0:
        tail_negative_modes.append({
            "mode": i,
            "q_idx": q_irr_idx + 1,
            "band": (i // Nq_irr) + 1,
            "w_re": w_re,
            "gamma": g_val
        })

    # imaginary mode check
    if w_re < 0:
        imaginary_modes.append({
            "mode": i,
            "q_idx": q_irr_idx + 1,
            "band": (i // Nq_irr) + 1,
            "w_re": w_re
        })

    # integration grid
    g_floor = max(g_val, 1e-4)
    dw = g_floor / 10.0

    w_min = max(0.001, w_qs + d_val - 100 * g_floor)
    #w_min = w_qs + d_val - 10 * g_floor
    w_max = w_qs + d_val + 100 * g_floor

    local_grid = np.arange(w_min, w_max, dw)

    mode_integral = 0.0
    spectral_weight = 0.0

    for w in local_grid:

        denom = (w - w_qs - d_val)**2 + g_val**2
        b_w = (1/np.pi) * (g_val / denom)

        spectral_weight += b_w * dw

        x = (w * THz_to_eV) / (kB * T)

        if x < 50:
            cv_kernel = kB * (x**2) * np.exp(x) / (np.exp(x) - 1)**2
            mode_integral += b_w * cv_kernel * dw

    # spectral loss check
    if spectral_weight < 0.90:
        spectral_loss_modes.append({
            "mode": i,
            "q_idx": q_irr_idx + 1,
            "band": (i // Nq_irr) + 1,
            "weight": spectral_weight
        })

    total_cv_weighted += mode_integral * weight
    integrated_modes_count += 1

    frequencies_qpa_d0.append(w_qs)
    cv_contributions_qpa_d0.append(mode_integral)


cv_per_cell = total_cv_weighted / Nq_total
cv_j_mol = cv_per_cell * 1.60218e-19 * 6.02214e23


print(f"Modes used (integrated): {integrated_modes_count}")
print(f"Cv of unitcell(mol of unitcell): {cv_j_mol:.4f} J/K/mol")

print(f"Imaginary modes: {len(imaginary_modes)}")
print(f"Spectral loss modes: {len(spectral_loss_modes)}")
print(f"Tail crossing modes: {len(tail_negative_modes)}")
```

```python
total_cv_weighted = 0.0
frequencies_qpa_d0 = []
cv_contributions_qpa_d0 = []

kB = 8.617333262e-5      # Boltzmann constant in eV/K
THz_to_eV = 0.004135667  # Conversion factor (h * 1e12)
T = 300.0                # Temperature in Kelvin
integrated_modes_count = 0

for i in range(num_total_modes):
    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]
    
    w_qs = w_bare[i]
    #if w_qs < 1e-10: continue
    
    # 1. Get Delta and Gamma for this specific irreducible mode
    #d_val = d_interps[i](w_qs)
    d_val = 0
    g_val = abs(g_interps[i](w_qs))
    
    # 2. Strict BQPA Lorentzian Integration
    g_floor = max(g_val, 1e-4) # prevent gamma=0
    dw = g_floor / 10.0 # mode-adative
    w_min = max(0.001, w_qs + d_val - 100 * g_floor)
    w_max = w_qs + d_val + 100 * g_floor
    local_grid = np.arange(w_min, w_max, dw)
    
    mode_integral = 0.0
    for w in local_grid:
        # Spectral Density b_qs(w)
        denom = (w - w_qs - d_val)**2 + g_val**2
        b_w = (1/np.pi) * (g_val / denom)
        
        # Heat Capacity Kernel
        x = (w * THz_to_eV) / (kB * T)
        if x < 50:
            cv_kernel = kB * (x**2) * np.exp(x) / (np.exp(x) - 1)**2
            mode_integral += b_w * cv_kernel * dw
    
    # Multiply the mode's contribution by its degeneracy weight
    total_cv_weighted += mode_integral * weight
    integrated_modes_count += 1

    frequencies_qpa_d0.append(w_qs)
    cv_contributions_qpa_d0.append(mode_integral)
    
# --- 3. Final Normalization by total BZ points ---
cv_per_cell = total_cv_weighted / Nq_total
cv_j_mol = cv_per_cell * 1.60218e-19 * 6.02214e23
print(f"Modes used (integrated): {integrated_modes_count}")
print(f"Cv of unitcell(mol of unitcell): {cv_j_mol:.4f} J/K/mol")
```

```python
total_cv_weighted = 0.0
frequencies_qpa = []
cv_contributions_qpa = []

kB = 8.617333262e-5      # Boltzmann constant in eV/K
THz_to_eV = 0.004135667  # Conversion factor (h * 1e12)
#T = 100.0                # Temperature in Kelvin

for i in range(num_total_modes):
    # Mapping the global index 'i' back to the irreducible q-point index
    # Since w_bare is (num_bands * Nq_irr), the q-index repeats every Nq_irr
    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]
    
    w_qs = w_bare[i]
    #if w_qs < 1e-10: continue
    
    # 1. Get Delta and Gamma for this specific irreducible mode
    d_val = d_interps[i](w_qs)
    #d_val = 0
    g_val = abs(g_interps[i](w_qs))
    
    # 2. Strict BQPA Lorentzian Integration
    g_floor = max(g_val, 1e-4) # prevent gamma=0
    dw = g_floor / 10.0 # mode-adative
    w_min = max(0.001, w_qs + d_val - 100 * g_floor)
    w_max = w_qs + d_val + 100 * g_floor
    local_grid = np.arange(w_min, w_max, dw)
    
    mode_integral = 0.0
    for w in local_grid:
        # Spectral Density b_qs(w)
        denom = (w - w_qs - d_val)**2 + g_val**2
        b_w = (1/np.pi) * (g_val / denom)
        
        # Heat Capacity Kernel
        x = (w * THz_to_eV) / (kB * T)
        if x < 50:
            cv_kernel = kB * (x**2) * np.exp(x) / (np.exp(x) - 1)**2
            mode_integral += b_w * cv_kernel * dw
    
    # Multiply the mode's contribution by its degeneracy weight
    total_cv_weighted += mode_integral * weight

    frequencies_qpa.append(w_qs)
    cv_contributions_qpa.append(mode_integral)
    
# --- 3. Final Normalization by total BZ points ---
cv_per_cell = total_cv_weighted / Nq_total
cv_j_mol = cv_per_cell * 1.60218e-19 * 6.02214e23

print(f"Cv of unitcell(mol of unitcell): {cv_j_mol:.4f} J/K/mol")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- 1. Data Preparation ---
# Convert list to NumPy array to enable element-wise float multiplication
freqs_array = np.array(frequencies_qpa)
# cv_qpa_array: Quasiparticle heat capacity contributions
cv_qpa_array = np.array(cv_contributions_qpa)

# --- 2. Calculate Absolute Deviation ---
# x_vals: Dimensionless energy parameter (hbar * omega / kB * T)
x_vals = (freqs_array * THz_to_eV) / (kB * T)

# cv_mode_harmonic: Heat capacity calculated using the harmonic kernel
# Using the standard Bose-Einstein derivative form
cv_mode_harmonic = kB * (x_vals**2) * np.exp(x_vals) / (np.exp(x_vals) - 1)**2

# Unit conversion: eV/K/mode -> J/K/mol
# Conversion factor: 1.60218e-19 (J/eV) * 6.02214e23 (atoms/mol)
conv_factor = 1.60218e-19 * 6.02214e23
cv_harmonic_conv = cv_mode_harmonic * conv_factor
cv_qpa_conv = cv_qpa_array * conv_factor

# deviation: Difference between Harmonic and QPA values
# A positive value represents the loss of heat capacity due to anharmonicity (damping)
deviation = cv_harmonic_conv - cv_qpa_conv

# --- 3. Histogram Plotting ---
plt.figure(figsize=(10, 6))
# Plotting the distribution of deviations across all phonon modes
counts, bins, _ = plt.hist(deviation, bins=50, color='teal', edgecolor='black', alpha=0.7)

# Statistical Indicators
mean_dev = np.mean(deviation)
plt.axvline(mean_dev, color='red', linestyle='--', linewidth=2, 
            label=f'Mean Deviation: {mean_dev:.3f} J/K/mol')

# Chart Formatting
plt.title(f'Distribution of $C_{{v}}$ Deviation (T = {T} K)', fontsize=14)
plt.xlabel('$\Delta C_v$ (Harmonic - QPA) [J/K/mol]', fontsize=12)
plt.ylabel('Frequency (Number of Modes)', fontsize=12)
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()

# --- 4. Identify Top Outliers ---
# Sorting indices by deviation in descending order
outlier_idx = np.argsort(deviation)[-5:][::-1]
print("--- Top 5 Anharmonic Modes (Largest Deviation) ---")
for i in outlier_idx:
    print(f"Frequency: {freqs_array[i]:.3f} THZ | Deviation: {deviation[i]:.3f} J/K/mol")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# --- 1. Reshape harmonic frequencies ---
# w_harmonic shape: (Nq_irr, num_bands)
w_harmonic = w_bare.reshape(num_bands, Nq_irr).T 

# --- 2. DOS Calculation Function ---
def calculate_harmonic_dos(frequencies, weights, energy_grid, sigma=0.05):
    """
    frequencies: (Nq_irr, num_bands)
    weights: (Nq_irr,) - Multiplicity of each q-point
    energy_grid: Frequency axis for the DOS plot
    sigma: Gaussian broadening width (THz)
    """
    dos = np.zeros_like(energy_grid)
    for i in range(len(weights)):
        for j in range(frequencies.shape[1]):
            # Add Gaussian contribution for each mode: weight * G(omega - w_ij)
            diff = energy_grid - frequencies[i, j]
            dos += weights[i] * np.exp(-0.5 * (diff / sigma)**2) / (sigma * np.sqrt(2 * np.pi))
    
    # Normalize by total number of q-points in full BZ
    return dos / np.sum(weights)

# --- 3. Setup Grid and Compute ---
# Create a frequency axis from 0 to slightly above max frequency
w_max = np.max(w_harmonic) * 1.05
energy_axis = np.linspace(0, w_max, 1000)

# Calculate DOS (You can adjust sigma for smoothness)
dos_harmonic = calculate_harmonic_dos(w_harmonic, degeneracies, energy_axis, sigma=0.04)

# --- 4. Plotting ---
plt.figure(figsize=(9, 6))
plt.plot(energy_axis, dos_harmonic, color='blue', linewidth=2, label='Harmonic DOS')
plt.fill_between(energy_axis, 0, dos_harmonic, color='blue', alpha=0.2)

plt.title('Harmonic Phonon Density of States (DOS)', fontsize=15)
plt.xlabel('Frequency (THz)', fontsize=12)
plt.ylabel('Density of States (States/THz)', fontsize=12)
plt.xlim(0, w_max)
plt.ylim(0, np.max(dos_harmonic) * 1.1)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.tight_layout()
plt.show()

print(f"Harmonic DOS plot complete.")
print(f"Frequency range: 0.0 to {np.max(w_harmonic):.3f} THz")
```

```python
import numpy as np

total_cv_full = 0.0
frequencies_bqpa = []
cv_contributions_bqpa = []

# Define global frequency bounds from the loaded spectrum file
global_w_min = spec_data[:, 0].min()
global_w_max = spec_data[:, 0].max()

print(f"Starting Full Frequency-Dependent BQPA integration...")

for i in range(num_total_modes):
    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]
    w_qs = w_bare[i]
    
    if w_qs < 1e-10: continue

    # Local step size based on broadening at resonance
    g_ref = abs(g_interps[i](w_qs))
    dw = max(g_ref, 1e-4) / 10.0
    
    # Integration grid for this mode
    w_grid = np.arange(global_w_min, global_w_max, dw)
    
    # Evaluate frequency-dependent Delta and Gamma
    d_ws = d_interps[i](w_grid)
    g_ws = np.abs(g_interps[i](w_grid))
    
    # Spectral Density b_qs(w)
    denoms = (w_grid - w_qs - d_ws)**2 + g_ws**2
    b_ws = (1/np.pi) * (g_ws / denoms)
    
    # Heat Capacity Kernel with Singularity Fix
    x = (w_grid * THz_to_eV) / (kB * T)
    cv_kernels = np.zeros_like(x)
    
    # Handle x=0 limit (kB) and general case
    m_zero = (x < 1e-7)
    m_finite = (x >= 1e-7) & (x < 50)
    
    cv_kernels[m_zero] = kB
    cv_kernels[m_finite] = kB * (x[m_finite]**2 * np.exp(x[m_finite])) / (np.exp(x[m_finite]) - 1)**2
    
    # Mode contribution (Area under b_w * Cv_kernel)
    mode_integral = np.sum(b_ws * cv_kernels) * dw
    total_cv_weighted = mode_integral * weight
    
    total_cv_full += total_cv_weighted
    frequencies_bqpa.append(w_qs)
    cv_contributions_bqpa.append(mode_integral)

# Final Normalization
cv_per_cell = total_cv_full / Nq_total
cv_j_mol = cv_per_cell * 1.60218e-19 * 6.02214e23

print(f"Full Frequency-Dependent Cv: {cv_j_mol:.4f} J/K/mol")
```

```python
import numpy as np
import matplotlib.pyplot as plt

gamma_fwhm_list = []
w_bare_actual = []

print("Extracting FWHM using Full BQPA Spectral Density...")

for i in range(num_total_modes):
    w_qs = w_bare[i]
    if w_qs < 0.1: continue

    # 1. Get the reference Gamma to set adaptive grid density
    g_ref = abs(g_interps[i](w_qs))
    dw = max(g_ref, 1e-4) / 10.0
    
    # Define frequency grid for this mode
    w_grid = np.arange(global_w_min, global_w_max, dw)
    
    # 2. FULL BQPA: Evaluate frequency-dependent Self-energy on the grid
    # This is the core of Full BQPA
    d_ws = d_interps[i](w_grid)
    g_ws = np.abs(g_interps[i](w_grid))
    
    # 3. Calculate the actual Spectral Density b_qs(w) for this mode
    denom = (w_grid - w_qs - d_ws)**2 + g_ws**2
    b_ws = (1.0 / np.pi) * (g_ws / denom)

    # --- FWHM Extraction Logic ---
    peak_idx = np.argmax(b_ws)
    max_b = b_ws[peak_idx]
    half_max = max_b / 2.0

    # Precise Left-wing (w_l) with Linear Interpolation
    left_part = b_ws[:peak_idx]
    if len(left_part) > 1:
        # Find first index crossing half_max from the left
        idx_l = np.where(left_part >= half_max)[0][0]
        if idx_l > 0:
            y1, y2 = left_part[idx_l-1], left_part[idx_l]
            w1, w2 = w_grid[idx_l-1], w_grid[idx_l]
            w_l = w1 + (w2 - w1) * (half_max - y1) / (y2 - y1)
        else:
            w_l = w_grid[idx_l]
    else:
        w_l = w_grid[0]

    # Precise Right-wing (w_r) with Linear Interpolation
    right_part = b_ws[peak_idx:]
    if len(right_part) > 1:
        # Find first index falling below half_max from the peak
        idx_r_rel = np.where(right_part <= half_max)[0][0]
        idx_r = idx_r_rel + peak_idx
        if idx_r_rel > 0:
            y1, y2 = right_part[idx_r_rel-1], right_part[idx_r_rel]
            w1, w2 = w_grid[idx_r-1], w_grid[idx_r]
            w_r = w1 + (w2 - w1) * (half_max - y1) / (y2 - y1)
        else:
            w_r = w_grid[idx_r]
    else:
        w_r = w_grid[-1]

    # Half-Width at Half-Maximum (HWHM)
    g_A = (w_r - w_l) / 2.0
    
    gamma_fwhm_list.append(g_A)
    w_bare_actual.append(w_qs)

# --- Visualization ---
plt.figure(figsize=(10, 6))
plt.scatter(w_bare_actual, gamma_fwhm_list, s=15, alpha=0.5, color='royalblue')
plt.title("Full BQPA Mode-resolved Gamma (FWHM Method)", fontsize=14)
plt.xlabel("Harmonic Frequency $\omega_0$ (THz)", fontsize=12)
plt.ylabel("Gamma $\Gamma$ (THz)", fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

```python
gamma_peak_list = []
w_bare_actual_B = []

print("Extracting Gamma using Method B (Peak Height)...")

for i in range(num_total_modes):
    w_qs = w_bare[i]
    if w_qs < 0.1: continue

    # 1. Full BQPA Spectral Density Calculation
    g_ref = abs(g_interps[i](w_qs))
    dw = max(g_ref, 1e-4) / 20.0
    w_grid = np.arange(global_w_min, global_w_max, dw)
    
    d_ws = d_interps[i](w_grid)
    g_ws = np.abs(g_interps[i](w_grid))
    
    denom = (w_grid - w_qs - d_ws)**2 + g_ws**2
    b_ws = (1.0 / np.pi) * (g_ws / denom)

    # 2. Method B: Extract Gamma from Peak Height
    # Formula: Gamma = 1 / (pi * b_max)
    b_max = np.max(b_ws)
    
    if b_max > 0:
        g_B = 1.0 / (np.pi * b_max)
    else:
        g_B = g_ref # Fallback if b_max is invalid

    gamma_peak_list.append(g_B)
    w_bare_actual_B.append(w_qs)

# --- Visualization: Method A vs Method B ---
plt.figure(figsize=(10, 6))
plt.scatter(w_bare_actual, gamma_fwhm_list, s=15, alpha=0.4, label='Method A (FWHM)', color='royalblue')
plt.scatter(w_bare_actual_B, gamma_peak_list, s=15, alpha=0.4, label='Method B (Peak Height)', color='orange')

plt.title("Comparison: Method A (FWHM) vs Method B (Peak Height)", fontsize=14)
plt.xlabel("Frequency $\omega_0$ (THz)", fontsize=12)
plt.ylabel("Gamma $\Gamma$ (THz)", fontsize=12)
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Results Containers ---
cv_sym_contributions = []
w_bare_actual = []
total_cv_sym_raw = 0.0
kB = 8.617333262e-5      # Boltzmann constant in eV/K
THz_to_eV = 0.004135667

print("Extracting FWHM & w_peak and calculating Symmetric Cv...")

for i in range(num_total_modes):
    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]
    w_qs = w_bare[i]
    
    if w_qs < 0.1: continue

    # 1. Prepare Adaptive Grid for the current mode
    g_ref = abs(g_interps[i](w_qs))
    dw_scan = max(g_ref, 1e-4) / 10.0
    w_grid_full = np.arange(global_w_min, global_w_max, dw_scan)
    
    # 2. Evaluate Full BQPA Spectral Density
    d_ws = d_interps[i](w_grid_full)
    g_ws = np.abs(g_interps[i](w_grid_full))
    b_ws = (1.0 / np.pi) * (g_ws / ((w_grid_full - w_qs - d_ws)**2 + g_ws**2))

    # 3. Precise Peak and FWHM Extraction (Using your interpolation logic)
    peak_idx = np.argmax(b_ws)
    max_b = b_ws[peak_idx]
    w_peak = w_grid_full[peak_idx] # Crude peak from grid
    half_max = max_b / 2.0

    # --- Left-wing (w_l) interpolation ---
    left_part = b_ws[:peak_idx]
    if len(left_part) > 1:
        idx_l = np.where(left_part >= half_max)[0][0]
        if idx_l > 0:
            y1, y2 = left_part[idx_l-1], left_part[idx_l]
            w1, w2 = w_grid_full[idx_l-1], w_grid_full[idx_l]
            w_l = w1 + (w2 - w1) * (half_max - y1) / (y2 - y1)
        else: w_l = w_grid_full[idx_l]
    else: w_l = w_grid_full[0]

    # --- Right-wing (w_r) interpolation ---
    right_part = b_ws[peak_idx:]
    if len(right_part) > 1:
        idx_r_rel = np.where(right_part <= half_max)[0][0]
        idx_r = idx_r_rel + peak_idx
        if idx_r_rel > 0:
            y1, y2 = right_part[idx_r_rel-1], right_part[idx_r_rel]
            w1, w2 = w_grid_full[idx_r-1], w_grid_full[idx_r]
            w_r = w1 + (w2 - w1) * (half_max - y1) / (y2 - y1)
        else: w_r = w_grid_full[idx_r]
    else: w_r = w_grid_full[-1]

    # Final Gamma (HWHM)
    g_fwhm = (w_r - w_l) / 2.0

    # 4. Define Integration Grid for Symmetric Model (+/- 100 * Gamma)
    # Grid density: Gamma / 10
    dw_sym = g_fwhm / 10.0
    w_start = max(1e-4, w_peak - 100 * g_fwhm)
    w_end = w_peak + 100 * g_fwhm
    w_grid_sym = np.arange(w_start, w_end, dw_sym)
    
    # 5. Perfect Symmetric Lorentzian (b_sym)
    denoms_sym = (w_grid_sym - w_peak)**2 + g_fwhm**2
    b_sym = (1.0 / np.pi) * (g_fwhm / denoms_sym)
    
    # 6. Heat Capacity Kernel 
    x = (w_grid_sym * THz_to_eV) / (kB * T)
    cv_kernels = np.zeros_like(x)
    m_zero = (x < 1e-7)
    m_finite = (x >= 1e-7) & (x < 50)
    
    cv_kernels[m_zero] = kB
    cv_kernels[m_finite] = kB * (x[m_finite]**2 * np.exp(x[m_finite])) / (np.exp(x[m_finite]) - 1)**2
    
    # 7. Integration & Weighting
    mode_integral_sym = np.sum(b_sym * cv_kernels) * dw_sym
    total_cv_sym_raw += (mode_integral_sym * weight)
    
    cv_sym_contributions.append(mode_integral_sym * 1.60218e-19 * 6.02214e23)
    w_bare_actual.append(w_qs)

# --- Final Unit Conversion (J/mol·K) ---
cv_per_cell_sym = total_cv_sym_raw / Nq_total
cv_j_mol_sym = cv_per_cell_sym * 1.60218e-19 * 6.02214e23

print("-" * 40)
print(f"Symmetric Model Cv: {cv_j_mol_sym:.4f} J/mol·K")
print("-" * 40)

plt.figure(figsize=(10, 6))
plt.scatter(w_bare_actual, cv_sym_contributions, s=15, color='pink', alpha=0.5)
if len(frequencies_bqpa) > 0:
    w_ref = np.linspace(min(frequencies_bqpa), max(frequencies_bqpa), 500)
    x_ref = (w_ref * THz_to_eV) / (kB * T)
    # C_v kernel calculation
    cv_harmonic = kB * (x_ref**2) * np.exp(x_ref) / (np.exp(x_ref) - 1)**2

    #cv_harmonic = np.zeros_like(x_ref)
    #small_x = x_ref < 1e-3
    #large_x = ~small_x
    #cv_harmonic[small_x] = kB
    #cv_harmonic[large_x] = kB * (x_ref[large_x]**2) * np.exp(x_ref[large_x]) / (np.exp(x_ref[large_x]) - 1)**2

    
    plt.plot(w_ref, cv_harmonic * 1.60218e-19 * 6.02214e23, color='red', linestyle='--', label='Harmonic Heat Capacity Kernel', linewidth=2)
plt.title(f"Cv from Symmetric Lorentzian Peak (Total: {cv_j_mol_sym:.3f} J/mol·K)")
plt.xlabel("Harmonic Frequency $\omega_0$ (THz)")
plt.ylabel("Mode Cv (J/K/mol)")
plt.grid(True, ls=':')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# --- 1. Setup Energy Grid ---
# Use the global bounds from your spectrum data
energy_axis = np.linspace(global_w_min, global_w_max, 1000)
dos_bqpa_full = np.zeros_like(energy_axis)

print("Calculating Full BQPA DOS (Frequency-dependent Spectral Density)...")

# --- 2. Accumulate Full Spectral Density for each mode ---
for i in range(num_total_modes):
    q_irr_idx = i % Nq_irr
    weight = degeneracies[q_irr_idx]
    w_qs = w_bare[i]
    
    # 1. Evaluate Delta(w) and Gamma(w) on the entire energy axis
    d_ws = d_interps[i](energy_axis)
    g_ws = np.abs(g_interps[i](energy_axis))
    
    # 2. Calculate the Full Spectral Density b_qs(w)
    # This reflects the true shape of the phonon state in an anharmonic potential
    denoms = (energy_axis - w_qs - d_ws)**2 + g_ws**2
    
    # Avoid division by zero for very stable modes
    g_ws_safe = np.where(g_ws < 1e-6, 1e-6, g_ws)
    b_ws = (1.0 / np.pi) * (g_ws_safe / denoms)
    
    # 3. Add to total DOS with degeneracy weight
    dos_bqpa_full += weight * b_ws

# --- 3. Final Normalization ---
dos_bqpa_full /= Nq_total

# --- 4. Plotting Comparison: Harmonic vs QPA vs BQPA ---
plt.figure(figsize=(12, 7))

# 1. Harmonic DOS (Reference)
plt.plot(energy_axis, dos_harmonic, 'k--', label='Harmonic DOS', alpha=0.3)

# 2. QPA DOS (From previous step - if you have the variable)
#if 'dos_qpa_varied' in locals():
#    plt.plot(energy_axis, dos_qpa_varied, color='green', label='QPA DOS (Single-point $\Sigma$)', alpha=0.5)

# 3. Full BQPA DOS
plt.plot(energy_axis, dos_bqpa_full, color='purple', linewidth=2.5, label='Full BQPA DOS (Freq-dependent $\Sigma$)')
plt.fill_between(energy_axis, 0, dos_bqpa_full, color='purple', alpha=0.15)

plt.title('Phonon DOS: Full BQPA Comparison', fontsize=16)
plt.xlabel('Frequency (THz)', fontsize=13)
plt.ylabel('Density of States', fontsize=13)
plt.xlim(0, global_w_max)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()

plt.tight_layout()
plt.show()
```
